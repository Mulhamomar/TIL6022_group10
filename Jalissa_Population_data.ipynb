Preliminary investigation: What Charts can be Useful for This Research and explain Why?

Since your dependent variable is bus ridership and your independent variables are median age, gender distribution, and population density, you’ll want charts that clearly show relationships and group differences.

1. Scatterplots: Use for: Median_Age vs. avg_boardings and pop_density vs. avg_boardings. Scatterplots can show the direct relationship between two variables. Adding a regression line helps visualize whether the relationship is linear (important for regression analysis).

2. Boxplots or Violin Plots: Use for: comparing avg_boardings across groups (e.g., census block groups with higher % male vs. higher % female). These plots are great for spotting group differences in ridership. They connect well with statistical tests such as the two-proportion Z-test.

3. Histograms / Density Plots: Use for: distributions of avg_boardings, Median_Age, and pop_density. They show the shape of the data (normal, skewed, outliers). This helps decide whether transformations (like log(pop_density)) are needed.

4. Correlation Heatmap (optional): Use for: all numerical variables (avg_boardings, Median_Age, %Male, %Female, pop_density). A heatmap gives a quick overview of correlations and helps check for multicollinearity between predictors.

5. Bar Charts (aggregated): Use for: average ridership per demographic class (e.g., dividing population density or age into quintiles). Bar charts simplify patterns into categories, making trends easier to communicate to a broader audience.
